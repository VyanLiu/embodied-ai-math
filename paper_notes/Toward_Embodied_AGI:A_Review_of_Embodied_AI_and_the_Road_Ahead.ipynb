# Toward Embodied AGI: A Review of Embodied AI and the Road Ahead

## Introduction to the Thesis
The core thesis of "Toward Embodied AGI: A Review of Embodied AI and the Road Ahead" centers on the evolution of artificial intelligence from disembodied foundation models to **Embodied Artificial General Intelligence (E-AGI)**. E-AGI represents AI systems capable of perceiving, reasoning, and dynamically acting within physical or highly realistic virtual environments. This transition is heavily driven by the integration of foundation models, which serve as the cognitive "brain" for embodied agents.

### The Integration of Foundation Models
Current embodied agents extensively leverage foundation models for common-sense reasoning, scene understanding, and zero-shot task planning. The field has progressed through the adaptation of various foundational architectures:
* **Large Language Models (LLMs):** such as GPT-4 (OpenAI, 2023) and LLaMA (Touvron et al., 2023), which provide profound semantic reasoning and high-level task planning.
* **Vision-Language Models (VLMs):** such as CLIP (Radford et al., 2021) and LLaVA (Liu et al., 2023), which align visual perception with linguistic concepts for spatial and semantic scene understanding.
* **Embodied and Vision-Language-Action (VLA) Models:** such as PaLM-E (Driess et al., 2023), RT-1 (Brohan et al., 2022), and RT-2 (Brohan et al., 2023), which directly map multimodal inputs to low-level robotic actions.

### Five Levels of Embodied AGI
To structure the progression toward ultimate autonomy, the paper defines 5 levels of E-AGI:
1. **Level 1: Single-task completion** - At this level, agents are capable of performing a single, well-defined task that is useful in industrial or everyday settings. Complex goals are decomposed into simpler subtasks by humans manually. A sufficiently robust body is needed to support its target tasks.
2. **Level 2: Compositional task completion** - Agents can perform a series of compositional tasks by decomposing high-level human instructions into sequences of simpler actions. They are limited by predefined tasks and skill libraries in a specific domain. supporting longer and more complex actions sequences is the main goal for Embodied AGI of this level.
3. **Level 3: Conditional general-purpose task completion** - Robots are capable of handling a wide range of task categories, demonstrating conditional generalization across tasks, environments, and human instructions. They don't need to rely on humans to perform a novel or open-ended task. At this level, robots adapt to the environmental changes and can response to humans in real-time. In the context of LLMs, this stage loosely resembles pre-trained foundation models(e.g., BERT, GPT-3, LLaMA 2) equipped with multitask fine-tuning or capable of few-shot prompting.
4. **Level 4: Highly general-purpose robots** robots in this level are capable of broad zero-shot generalization across unseen tasks, marking true general purpose capability. The high-level tasks that a robot can perform are not only real-time processing, but also strong multimodel comprehension and reasoning abilities (e.g., across language, audio and vision), ensuring sophisticated communication and interaction with humans. For LLM analogy, The general-purpose LLMs equipped with strong reasoning capabilities, such as o1 and deepseek r1 are in Level 4.
5. **Level 5: All-purpose robots** - L5 represents the ultimate goal of Embodied AGI: developing genuinely all-purpose robotic agents capable of meeting a wide spectrum of human daily needs. These robots integrate a deep understanding of physical laws and human emotional and social dynamics, process all modalities seamlessly in real-time. They exhibit distinctly human-like cognitive behaviors, including self-awareness, social connection understanding, procedural memory, and memory reconsolidation.

### Four Core Dimensions to Assess Capabilities
Evaluating an embodied agent's progression through these levels relies on assessing four core dimensions:
1. **Omnimodal capabilities:** the ability to process a full spectrum of information modalities
2. **humanoid cognitive abilities:** for nuance social comprehension and human-like learning mechanisms, including self-awareness, social connection understanding, procedural memory, and memory reconsolidation.
3. **real-time responsiveness:** the capability to conduct swift, accurate actions and duplex interaction.
4. **generalization:** The capability to adapt to open-ended environment and real-world tasks.

**Our Current State:** The field of embodied AI is currently situated at a transitional phase between **Level 1 and Level 2**, with cutting-edge research making early strides into **Level 4**. While task-specific automation is well-established, modern foundation models are rapidly enabling multi-task capabilities and sparking the first signs of cross-embodiment generalization.

### Review in each level proposed by the paper
#### Level 1 ~ Level 2
* **end-to-end**: directly processing visual and textual inputs to generate actions via next token prediction() or diffusion-based methods, e.g., VLA models.
* **plan-and-act**: first utilize VLMs or LLMs to interpret multimodal inputs and then perform high-level planning and task decomposition, generating intermediate control signals, such as executable code, function calls, or verbal instructions.
* **others**: latent-space planning

Many can reliably complete single tasks with demonstrated to previously unseen environments and conditions.
* **GraspVLA**: a multimodel VLA can generalize grasping across various lighting conditions, backgrounds, distractions, and object heights.
* **State-of-the-art robotic systems**: such as Helix, not only display robust generalization within specific task types(e.g., picking diverse objects), but can also handle a wide range of dexterous indoor tasks.
* **\PI_\text{0.5}:** partly address diverse task categories though combined pre-training(e.g. mobile and non-mobile tasks), yet their applications still largely focus on environmental generalization rather than genuine task diversity.

**Challenges to and beyond L3:**
1. **Lack of comprehensive joint-modal capabilities:** True embodied intelligence necessitates full-spectrum multimodal perception(e.g., understanding human speech with emotion and sentiment; listening to environmental audio inputs from microphone devices in addition to text console and imagery camera) and multimodal responses, including real-time acoustic speech feedback.
2. **Insufficient humanoid cognition:** Fully capable embodied agents must excel in reasoning and conversational intelligence, akin to sophisticated chatbots, and demonstrate alignment with human preferences and ethical values.
3. **Limited real-time responsiveness:** Most current embodied AI systems operate in a simi-duplex manner: receiving and processing instructions fully before acting, making them struggle for dynamic environments where conditions or instructions change rapidly.
4. **Restricted generalization:** there are still a wide range of cross-environment generation scenarios that current models struggle to handle, a typical example being the invariance to spatial transformations.

#### Level 3 ~ Level 5

**1. Omnimodal Capabilities:** 
The core is the "general-purpose" nature, achieved only through comprehensive omnimodal capabilities beyond vision and language. Current researches are bimodal foundational models such as visual-language (Cheang et al. 2024) and audio-language models (Zeng et al.2024). Tri-modal models (e.g., vision-language-audio) have garnered considerable interest (Gemini 2024; Xu et al. 2025). However, incorporating additional modalities like actions and environmental sensing for embodied agents remains largely uncharted.
* **Challenges**: 
  * a. Modality conflicts, which impose high demands on model capacity (Aghajanyan et al. 2023). 
  * b. Cascading errors and alignments issues arising from modality-specific modules and heterogeneous data distributions (Tong et al. 2024).
* **Solutions**: 
  * a. Parallel understanding-inference-generation architectures to reduce time cost. 
  * b. More advanced multimodal pretraining paradigms that improve the collaboration of modality-specific modules or inherently support multimodal comprehension.

**2. Humanoid Cognitive Abilities:** 
Mimicking essential learning mechanisms of the human neural brain (Liu et al. 2025) potentially enhances the capability of embodied agents. Additionally, a humanoid understanding of self and social connections significantly improves the quality of human-robot interaction.
* **Self-awareness:** The foundation of higher cognitive functions (Liang et al.2024). This awareness should be lifelong, dynamic, and stateful.
* **Social connection understanding:** Such awareness helps an AI system comprehend its roles, responsibilities, and character (Chen et al.2024). It should also be lifelong, dynamic, and stateful.
* **Procedural memory:** Essential for overcoming domain shifts and addressing catastrophic forgetting (Cavaco et al.2024).
* **Memory reconsolidation:** Facilitates long context learning (Wang et al.2024) by extending the context window (Dao 2023) and optimizing positional encoding (Sun et al.2023).

**3. Real-time Interaction:**
Transitioning from turn-based (semi-duplex) to full-duplex communication and action. Agents in levels 3-5 must continuously stream and process sensory inputs (visual, auditory, tactile) while executing and dynamically adjusting physical actions. This involves low-latency edge inference, the ability to be interrupted by humans mid-task, and instantaneous adaptation to rapidly changing physical environments without pausing to "re-plan."

**4. Generalization to Open-ended Tasks:**
Moving beyond predefined skill libraries and controlled environments to handle novel, unstructured real-world challenges. This dimension requires:
* **Zero-shot physical reasoning:** Deriving action strategies for tasks never seen during training based on physical common sense.
* **On-the-fly skill synthesis:** Combining basic atomic movements into complex, novel behaviors to solve unprecedented problems.
* **Cross-embodiment adaptability:** Transferring learned cognitive and physical skills seamlessly across different robotic morphologies (e.g., from a robotic arm to a quadruped or a humanoid) in an open-world setting.

### Limitations and Obstacles
Despite rapid advancements, developers of Embodied AGI face several critical challenges:
* **Data Scarcity for Embodiment:** Unlike the massive text and image datasets available for LLMs/VLMs, high-quality, diverse physical interaction data (robotic trajectories, multimodal sensor logs) is extremely costly and slow to collect.
* **The Sim-to-Real Gap:** While simulators provide safe and scalable training environments, discrepancies in physics modeling, sensor noise, and unforeseen physical dynamics make it difficult to transfer learned policies directly to real-world robots.
* **Hardware Constraints and Edge Computing:** Deploying giant foundation models requires massive computational power. Executing these models in real-time on physical robots (which have strict constraints on payload, battery, and compute) remains a significant bottleneck.
* **Safety, Robustness, and Alignment:** In the physical world, mistakes can cause actual damage. Ensuring agents act safely in unstructured, unpredictable human environments and align with human intentions is far more complex than digital safety filters.
* **Cross-Embodiment Generalization:** Morphological differences among robots (e.g., varying joint limits, sensor placements, and kinematics) make it highly challenging to create a "universal brain" that seamlessly operates any physical body.
* **Lack of Unified Benchmarks:** Unlike standardized NLP or CV benchmarks, evaluating embodied agents is fragmented, requiring complex physical setups or heavily customized simulations that are hard to standardize.

In summary, while current foundation models exhibit "sparks" of AGI in the digital realm, realizing true Embodied AGI demands mastering these four core dimensions and bridging the gap between abstract algorithmic cognition and generalizable physical execution.
