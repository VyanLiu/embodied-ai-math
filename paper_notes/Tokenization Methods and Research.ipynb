# 🔤 Tokenization Methods: From Words to Subwords

Tokenization is the critical first step in Natural Language Processing (NLP). It defines how a continuous stream of text is broken down into discrete units (tokens) that a machine learning model can process. This note explores the evolution of tokenization, from simple word-level splits to the sophisticated subword algorithms used by modern Transformers like GPT-4 and Llama-3.

### 1. The Hierarchy of Tokenization

| Method | Unit | Pros | Cons |
| :--- | :--- | :--- | :--- |
| **Word-Level** | Space-separated words | High semantic meaning; intuitive. | Large vocab; OOV (Out-of-Vocabulary) issues; morphologically blind. |
| **Character-Level** | Individual letters | Small vocab; no OOV; captures spelling. | Long sequences; low semantic density; computationally expensive. |
| **Subword-Level** | Meaningful word fragments | **The Goldilocks Zone**: Balances vocab size and sequence length. | Requires a training phase (building the vocab). |

### 2. Modern Subword Algorithms

#### 🔹 Byte-Pair Encoding (BPE)
*   **Concept:** Iteratively merges the most frequent pair of adjacent characters or sequences.
*   **Used In:** GPT-2/3/4, Llama, RoBERTa.
*   **Paper:** *Neural Machine Translation of Rare Words with Subword Units* (Sennrich et al., 2016).

#### 🔹 WordPiece
*   **Concept:** Similar to BPE but merges based on maximum likelihood (how much a merge increases the probability of the training data) rather than just frequency.
*   **Used In:** BERT.
*   **Paper:** *Japanese and Korean Voice Search* (Schuster and Nakajima, 2012).

#### 🔹 Unigram
*   **Concept:** Starts with a massive vocabulary and iteratively removes tokens that contribute least to the likelihood of the training corpus.
*   **Used In:** T5, ALBERT.
*   **Paper:** *Subword Regularization: Improving Neural Network Translation Models* (Kudo, 2018).

### 3. Essential Paper List

1.  **[BPE]** [Neural Machine Translation of Rare Words with Subword Units](https://arxiv.org/abs/1508.07909) (2016) - *The foundational paper for subword tokenization.*
2.  **[WordPiece]** [Japanese and Korean Voice Search](https://ieeexplore.ieee.org/document/6424251) (2012) - *Introduced the algorithm later popularized by BERT.*
3.  **[Unigram]** [Subword Regularization: Improving Neural Network Translation Models](https://arxiv.org/abs/1804.10959) (2018) - *Introduced probabilistic tokenization.*
4.  **[SentencePiece]** [SentencePiece: A simple and language independent subword tokenizer](https://arxiv.org/abs/1808.06226) (2018) - *A framework that treats whitespace as a normal symbol.*
5.  **[BBPE]** [Language Models are Few-Shot Learners (GPT-3)](https://arxiv.org/abs/2005.14165) (2020) - *Standardized Byte-level BPE to handle any character without UNK tokens.*

### 4. Code Implementation (BPE Example)
In our project, we transitioned from word-level to **Byte-Pair Encoding** in code_demo/train_en_fr.py to match the original Transformer performance. Here is how we implemented the BPE vocabulary:

```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

class Vocabulary:
    """Subword-level vocabulary using Byte-Pair Encoding (BPE)."""
    def __init__(self, max_vocab_size: int = 32000):
        self.tokenizer = Tokenizer(BPE(unk_token="<UNK>"))
        self.tokenizer.pre_tokenizer = Whitespace()
        self.trainer = BpeTrainer(
            vocab_size=max_vocab_size, 
            special_tokens=["<PAD>", "<BOS>", "<EOS>", "<UNK>"]
        )

    def build(self, sentences):
        self.tokenizer.train_from_iterator(sentences, self.trainer)
        
    def encode(self, text):
        return self.tokenizer.encode(text).ids
```
