# 🗺️ The Dual-Track Professional Study Plan (Refactored)

## 🧠 Core Principles & Tools

*   **The 80/20 Rule:** 80% of your time is on the **Core Practical** track (building things). 20% is on the **Math Foundation** track (understanding why they work).
*   **AI as Your Socratic Tutor:** Use JetBrains AI / Gemini Pro as a constant companion to summarize papers, explain code, and clarify math.
*   **The "Imagination" Workflow:** For any new architecture, follow this cycle:
    1.  **Concept:** Read about it in "Dive into Deep Learning" (d2l.ai) to get the core idea and a simple implementation.
    2.  **Intent:** Read the original paper on arXiv to understand why the authors created it.
    3.  **Craftsmanship:** Read the implementation in the `timm` library on GitHub to see how a world-class engineer builds it for production.
*   **The Math Log:** Maintain a single document where you log every new mathematical concept, its intuitive explanation, and a link to the code where you saw it in action.

## 🗓️ Month 1: Foundations & Mastering the "Paper-to-Code" Workflow
> **Objective:** Build a repeatable process for turning cutting-edge research into working code, mastering the "brain" of an embodied agent.

### 🛠️ Track 1: Core Practical (80% of your time)

#### Week 1: Habit Building & Surveying the Landscape
*   **Actions:**
    *   **Community:** Create a dedicated Twitter account. Follow 10 researchers from Google DeepMind and Meta AI. Follow the accounts for CoRL and CVPR.
    *   **News:** Subscribe to the "Import AI" and "The Batch" newsletters.
    *   **Concepts:** Watch Stanford's CS231n lectures on "Vision and Language" and "Deep Reinforcement Learning."
*   **Outcome:** A diagram of the agent-environment loop in your notes. You are now receiving a weekly feed of the latest AI news and research.

#### Weeks 2-4: The "Paper-to-Code" Gauntlet with Robotics Transformer (RT-1)
*   **Actions:**
    *   **Find the Paper & Code:** Use paperswithcode.com to find the RT-1 paper and its official GitHub repository.
    *   **AI-Assisted First Pass:** Use your AI to summarize the paper's abstract and "Methods" section.
    *   **Run the Official Code:** Your primary goal is to get the official demo script running successfully in its own environment.
    *   **Guided Re-implementation:** Apply the "Imagination" Workflow.
        *   Read the d2l.ai chapter on Attention Mechanisms.
        *   Re-read the RT-1 paper's methods section.
        *   In a new file, re-implement the `forward()` pass of the model, using the official code as your guide.
*   **Resources:** paperswithcode.com, arXiv, GitHub, d2l.ai, your AI assistant.
*   **Outcome:** A working script where you have partially re-implemented a major robotics model. You have a proven, repeatable workflow for tackling complex papers.

### 📐 Track 2: Math Foundation (20% of your time)
*   **Actions:**
    1.  Begin your "Math Log."
    2.  Work through the introductory sections of "Dive into Deep Learning" (d2l.ai).
*   **Resources:** d2l.ai.
*   **Focus Chapters (d2l.ai):**
    *   "Introduction," "Preliminaries" (for setup).
    *   "Linear Algebra," "Calculus" (for Autograd), "Automatic Differentiation."
    *   "Linear Neural Networks," "Multilayer Perceptrons (MLPs)."
*   **Outcome:** Your Math Log explains "Tensor," "Gradient," "Softmax," and "Cross-Entropy Loss," with links to the simple MLP code in d2l.ai.

## 🗓️ Month 2: The Body and The World
> **Objective:** Gain proficiency in a standard industry simulator and connect a policy "brain" to a robot "body."

### 🛠️ Track 1: Core Practical (80%)

#### Weeks 5-6: Simulator Mastery
*   **Actions:**
    *   Install and configure NVIDIA Isaac Sim.
    *   Complete the core tutorials: "Hello World," "UI Basics," and "Controlling a UR10 Robot Arm."
*   **Resources:** Isaac Sim Documentation.
*   **Outcome:** You can programmatically load a robot, read its sensors, and command its joints.

#### Weeks 7-8: Implement a Core Manipulation Task
*   **Actions:**
    *   **Find a Project:** Search arXiv for a simple manipulation paper, e.g., "Isaac Sim block stacking."
    *   **Architect:** Design the software structure for your project (modules for Environment, Agent, Policy, etc.) before writing implementation code.
    *   **Implement:** Build the system. The Policy will likely have a vision backbone (e.g., a ResNet).
    *   **Apply the "Imagination" Workflow:** After you get your vision model working, find the same architecture (e.g., `resnet.py`) in the `timm` GitHub repository. Read the professional implementation and note the differences and engineering patterns.
*   **Resources:** arXiv, your DDD knowledge, `timm` GitHub repository.
*   **Outcome:** A GitHub repository with a well-structured project where a robot in Isaac Sim can perform a simple, learned task.

### 📐 Track 2: Math Foundation (20%)
*   **Actions:** Focus on the math behind the vision models you are now using.
*   **Resources:** d2l.ai.
*   **Focus Chapters (d2l.ai):**
    *   "Deep Learning Computation" (understanding layers and blocks).
    *   "Convolutional Neural Networks" (LeNet, AlexNet, VGG, NiN, ResNet).
*   **Outcome:** Your Math Log now details "Convolution," "Pooling," and "Residual Connection," with links to the ResNet block code in d2l.ai and timm.

## 🗓️ Month 3: Synthesis & The Leadership Capstone
> **Objective:** Integrate advanced concepts into a portfolio piece that proves your ability to lead an Embodied AI project.

### 🛠️ Track 1: Core Practical (80%)

#### Week 9: Learn about Vision-Language-Action (VLA) Models
*   **Actions:** Use your AI-assisted workflow to read the blog posts and abstracts for RT-2 and SayCan.
*   **Goal:** Understand how these models map a language command and an image observation to a robot action.

#### Weeks 10-12: The Capstone Project: "Language-Commanded Manipulation"
*   **Actions:**
    *   **Architect:** Design a system in Isaac Sim for the command: "push the red block to the left." Draw the full component diagram.
    *   **Integrate, Don't Train:** Your Policy module will use a pre-trained CLIP model from the Hugging Face Hub.
    *   **Engineer the Logic:** Write the code that:
        1.  Takes an image from the simulator's camera.
        2.  Uses CLIP to find the coordinates of the object matching the text command ("red block").
        3.  Translates these coordinates into a sequence of robot arm commands.
*   **Resources:** Your project from Month 2, Hugging Face Hub (transformers library), your architecture design skills.
*   **Outcome:** A polished GitHub repository with a video demo and a README.md that clearly explains the architecture. This is your proof of concept for a team lead role.

### 📐 Track 2: Math Foundation (20%)
*   **Actions:** Focus on the math behind the sequence and language models you are now using.
*   **Resources:** d2l.ai.
*   **Focus Chapters (d2l.ai):**
    *   "Recurrent Neural Networks (RNNs)."
    *   "Modern Recurrent Neural Networks" (GRU, LSTM).
    *   "Attention Mechanisms."
*   **Outcome:** Your Math Log now explains the core concepts behind Transformers. You understand why attention is so powerful for processing both language and sequences of actions.
