# The Dual-Track Professional Study Plan

## 🧭 Guiding Principles
*   **Practicality First (The 80/20 Rule):** 80% of your time will be dedicated to hands-on engineering and paper implementation. 20% will be for foundational math. The practical work will provide the context that makes the theory meaningful.
*   **AI as a Socratic Tutor:** Use your JetBrains AI and Gemini Pro relentlessly. Treat it as a personal tutor that can summarize papers, explain code, and clarify mathematical concepts 24/7.
*   **From Paper to Product:** Your primary goal for each project is not to invent a new algorithm, but to create a well-engineered, working system from existing research. This is the skill of a team lead.
*   **The Math Log:** Maintain a digital document (e.g., in Notion, Obsidian, or a simple text file) where you log every new mathematical concept you learn, along with an intuitive explanation and a link to the practical code where you saw it applied.

## 📅 Month 1: Foundations & Mastering the "Paper-to-Code" Workflow
**Objective:** Overcome your primary weakness by creating a repeatable process for dissecting and implementing research papers. Master the "brain" of an embodied agent.

### 🛤️ Track 1: Core Practical (80% of your time)

#### Week 1: Survey the Landscape
*   **Actions:**
    1.  Watch Stanford's CS231n lectures on "Vision and Language" and "Deep Reinforcement Learning."
    2.  Read survey blog posts like "A Guide to Embodied AI" on the Hugging Face blog.
*   **Resources:** YouTube, Hugging Face Blog.
*   **Outcome:** A diagram in your notes illustrating the full agent-environment loop (State, Observation, Action, Policy, Reward) and a written definition for each term.

#### Weeks 2-4: The "Paper-to-Code" Gauntlet with Robotics Transformer (RT-1)
*   **Actions:**
    1.  **AI-Assisted First Pass:** Use your AI to summarize the RT-1 paper, focusing on its core contribution and the "Methods" section.
    2.  **Diagram-Driven Study:** Print out Figure 2 (the architecture diagram). This is your primary blueprint.
    3.  **Run the Official Code:** Find the official RT-1 GitHub repository. Your goal is to successfully set up the environment and run their provided demo script. This is a critical first win.
    4.  **Code Dissection:** Place the code for the model architecture side-by-side with the paper's diagram. Trace the data flow and map code blocks to diagram components.
    5.  **Guided Re-implementation:** In a new file (`my_rt1_study.py`), re-implement the `forward()` pass of the RT-1 model. Use the official code as a reference, but type every line yourself. Use your AI to explain any line you don't understand.
*   **Resources:** "RT-1" paper on arXiv, official GitHub repo, your AI assistant.
*   **Outcome:** A working, documented script where you have partially re-implemented a major robotics model. You now have a proven workflow for tackling complex papers.

### 📚 Track 2: Math Foundation (20% of your time)
*   **Actions:**
    1.  Read and work through Part I of the "Flower Book" (Deep Learning).
    2.  Start your "Math Log."
*   **Resources:** *Deep Learning* by Goodfellow, Bengio, Courville.
*   **Focus Chapters:**
    *   **Chapter 2: Linear Algebra** (Focus on dot products, matrix multiplication, and the concept of a vector space).
    *   **Chapter 3: Probability and Information Theory** (Focus on probability distributions, conditional probability, and the chain rule).
*   **Outcome:** Completion of the core math chapters. Your Math Log will have entries for concepts like "Softmax," "Cross-Entropy," and "Dot-Product Attention," linking them to their use in the RT-1 code you just studied.

## 📅 Month 2: The Body and The World
**Objective:** Gain proficiency in a standard industry simulator and connect a policy "brain" to a robot "body" to perform a core robotics task.

### 🛤️ Track 1: Core Practical (80%)

#### Weeks 5-6: Simulator Mastery
*   **Actions:**
    1.  Install and configure **NVIDIA Isaac Sim**. Given your hardware, this is the best choice for manipulation tasks.
    2.  Complete the "Hello World," "Adding a Simple Prim," and "UR10 Robot" tutorials in the Isaac Sim documentation.
*   **Resources:** Isaac Sim Documentation.
*   **Outcome:** You can programmatically load a robot into a scene, read its sensor data, and command its joints to move.

#### Weeks 7-8: Implement a Core Manipulation Task
*   **Actions:**
    1.  Apply your "Paper-to-Code" workflow to a simpler manipulation paper, such as one involving block stacking or reaching a target.
    2.  **Engineering Challenge:** Design the software architecture for your project. Create distinct Python modules for the **Environment** (your Isaac Sim wrapper), the **Agent** (which holds the policy), and the **Policy** (the model you're implementing).
    3.  Implement the connection between your policy's output (e.g., a target end-effector position) and the Isaac Sim commands needed to achieve it.
*   **Resources:** Papers on ArXiv (search "Isaac Sim block stacking"), your design skills (DDD).
*   **Outcome:** A GitHub repository containing a well-structured project where a robot in Isaac Sim can perform a simple, learned task.

### 📚 Track 2: Math Foundation (20%)
*   **Actions:** Continue with the Flower Book, focusing on the mechanics of training.
*   **Resources:** *Deep Learning* book.
*   **Focus Chapters:**
    *   **Chapter 5: Machine Learning Basics** (Understand capacity, overfitting, underfitting, and regularization like weight decay).
    *   **Chapter 8: Optimization for Training Deep Models** (Focus on Stochastic Gradient Descent (SGD) and Adam. Understand intuitively why momentum and adaptive learning rates are important).
*   **Outcome:** Your Math Log now explains concepts like "Gradient," "Learning Rate," "Loss Function," and "Optimizer." You understand what happens when you call `optimizer.step()`.

## 📅 Month 3: Synthesis & The Leadership Capstone
**Objective:** Integrate advanced concepts into a well-engineered system, creating a portfolio piece that demonstrates your ability to lead an Embodied AI project.

### 🛤️ Track 1: Core Practical (80%)

#### Week 9: Learn about Vision-Language-Action (VLA) Models
*   **Actions:** Use your AI-assisted workflow to read the blog posts and abstracts for RT-2 and SayCan.
*   **Goal:** Understand how these models map a natural language command and an image observation to a robot action.

#### Weeks 10-12: The Capstone Project: "Language-Commanded Manipulation"
*   **Actions:**
    1.  **Architect:** Design a system in Isaac Sim where a user can type a command like "push the red block to the left." Draw the full system diagram before writing any code.
    2.  **Integrate, Don't Train:** Use a powerful pre-trained model. Your Policy module will use **CLIP** to identify the target object. It will take the image from the simulator's camera, get CLIP embeddings for image patches and the text command (e.g., "red block"), and find the coordinates of the best match.
    3.  **Engineer the Logic:** Write the code that translates the identified object's coordinates into a sequence of robot arm commands (e.g., "move above the block," "move down," "move left").
*   **Resources:** Your Isaac Sim project from Month 2, CLIP from Hugging Face, your architecture design skills.
*   **Outcome:** A polished GitHub repository with a video demo. The `README.md` clearly explains the architecture you designed. This is your proof of concept for a team lead role.

### 📚 Track 2: Math Foundation (20%)
*   **Actions:** Dive into the specific architectures you've been using all along.
*   **Resources:** *Deep Learning* book.
*   **Focus Chapters:**
    *   **Chapter 9: Convolutional Networks** (Understand convolution, pooling, and why they are effective for images).
    *   **Chapter 10: Sequence Modeling: Recurrent and Recursive Nets** (Focus on the core ideas of handling sequential data, which is crucial for action sequences).
*   **Outcome:** Your Math Log now details the core building blocks of the models you've implemented. You understand the "why" behind the architectures.
