# Lecture 8: Attention and Transformers

### 1. From RNNs to Attention
*   **The Problem (RNN Bottleneck):** Traditional Seq2Seq RNNs force the encoder to compress the entire input sequence into a single, fixed-length context vector. This leads to information loss, especially for long sequences.
*   **The Solution (Attention):** Instead of a single vector, allow the decoder to "attend" to different parts of the encoder's hidden states at every time step.
    *   **Alignment Scores:** Calculate similarity between decoder state and encoder states.
    *   **Softmax:** Convert scores to probabilities.
    *   **Weighted Sum:** Create a context vector as a weighted average of encoder states.

### 2. The General Attention Mechanism (Q, K, V)
Attention is generalized as looking up information in a database:
*   **Query ($Q$):** What we are looking for (e.g., current decoder state).
*   **Key ($K$):** How input features are indexed (e.g., encoder states).
*   **Value ($V$):** The actual content (e.g., encoder states).

$$ \text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right)V $$

### 3. The Transformer Architecture
Introduced in *"Attention Is All You Need"* (Vaswani et al., 2017), discarding recurrence entirely.

#### Key Components:
*   **Self-Attention:** Each token in the sequence looks at every other token to compute its representation. It captures global context in a single layer.
*   **Multi-Head Attention:** Multiple attention "heads" run in parallel, allowing the model to focus on different subspaces (e.g., syntax vs. semantics).
*   **Positional Encoding:** Since attention is permutation-invariant (order doesn't matter), explicit position information (sine/cosine functions) is added to embeddings.
*   **Layer Norm & Residuals:** Essential for training deep stacks of Transformer blocks.

### 4. Why Transformers Won
1.  **Parallelization:** Unlike RNNs (sequential $t \to t+1$), Transformers process the whole sequence simultaneously, enabling massive scaling on GPUs.
2.  **Long-Term Dependencies:** The path length between any two tokens is $O(1)$, making it easier to learn relationships between distant words.
