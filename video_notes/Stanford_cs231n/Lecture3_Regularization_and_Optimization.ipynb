# Lecture 3: Regularization and Optimization

### 📉 General Loss Function

> 🎯 **Goal:** Find weights $W$ that minimize "badness". We want a model that fits the data well (**Fidelity**) but remains simple enough to generalize (**Simplicity**).

#### 1. The Master Equation

The total loss $L$ is a sum of two competing terms:

$$
L(W) = \underbrace{\frac{1}{N} \sum_{i=1}^N L_i(f(x_i, W), y_i)}_{\text{Data Loss (Fidelity)}} + \underbrace{\lambda R(W)}_{\text{Regularization (Simplicity)}}
$$

#### 2. Component Breakdown

| Feature | 📊 Data Loss | 🧩 Regularization |
| :--- | :--- | :--- |
| **Role** | **Fidelity** (Fit the training data) | **Simplicity** (Prevent overfitting) |
| **Depends on** | Data ($x_i, y_i$) and Weights ($W$) | Only Weights ($W$) |
| **Goal** | Minimize prediction error vs ground truth. | Encourage small/diffuse weights. |
| **Examples** | Cross-Entropy (Softmax), Hinge Loss (SVM) | $L_2$ (Weight Decay), $L_1$ (Lasso) |
| **Why?** | To learn the patterns in the data. | To apply Occam's Razor & generalize. |

#### 3. The Hyperparameter $\lambda$

$\lambda$ is the **control knob** for the trade-off.

*   **High $\lambda$** $\to$ Strong Regularization $\to$ Simple model (Risk: Underfitting)
*   **Low $\lambda$** $\to$ Weak Regularization $\to$ Complex model (Risk: Overfitting)

#### 4. Common Regularizers

*   **$L_2$ (Ridge/Weight Decay):** $R(W) = \sum W^2$. Penalizes peaky weights; prefers diffuse weight vectors.
*   **$L_1$ (Lasso):** $R(W) = \sum |W|$. Encourages sparsity (drives some weights to exactly zero).
*   **Elastic Net:** Combines $L_1$ and $L_2$.

#### 5. Optimization Landscape

Once the loss function $L(W)$ is defined, we use **Optimization** to find the minimal $W$.

*   **📉 Gradient Descent:** Calculate $\nabla_W L$ using *all* data, update $W$. Slow but precise.
*   **🎲 Stochastic Gradient Descent (SGD):** Update $W$ using *one* example at a time. Fast, noisy.
*   **📦 Mini-batch SGD:** Update $W$ using a small batch (e.g., 32, 64). The industry standard.
*   **🚀 Adam:** Adaptive step sizes; generally the default choice for starting deep learning projects.
