# How to represent experience as data?

state **$s_\text{t}$** - the state of the "world" at time **$t$**

observation **$o_\text{t}$** - what the agent observes at time **$t$**(only used when missing information)

action **$a_\text{t}$** - the decision taken at time **$t$**

trajectory **$\tau$** - sequences of states/observations and actions taken by the agent, denoted by **$\tau = (s_0, o_0, a_0, s_1, o_1, a_1, \ldots, s_{T-1}, o_{T-1}, a_{T-1}, s_T)$**

reward function **$r(s, a)$** - the reward received by the agent for taking action **$a$** in state **$s$**

## Some examples here

1. **robot example**

state **$s$** - RGB images, joint positions, joint velocities

action **$a$** - commanded next joint position

trajectory **$\tau$** - 10-sec sequence of camera, joint readings, controls at 20Hz **$(s_\text{1}, a_\text{1}, s_\text{2}, a_\text{2}, \ldots, s_\text{T}, a_\text{T}) T = 200$**

reward **$r(s, a)$** = 1 if the towel is on the hook in state **$s$**, 0 otherwise

2. **chatbot example**

observation **$o$** - the user's most recent message

action **$a$** - the chatbot's next message

trajectory **$\tau$** - variable length conversation trace **$(o_\text{1},a_\text{1},o_\text{2},a_\text{2},\ldots,o_\text{T},a_\text{T})$**

reward **$r(s, a)$** = 1 if the user gives upvote, -10 if the user downvotes, 0 if no user feedback


# How to represent behavior with a neural network?

State **$s$** $\to$ neural network $\to$ **$a$** by train **$\pi_\theta(a|s)$**

Observe state **$s_\text{t}$**

Take action **$a_\text{t}$** by sampling from policy **$\pi_\theta(\cdot|s_\text{t})$**

Observe next state **$s_\text{t+1}$** sampled from unknown world dynamics **$p(\cdot|s_\text{t}, a_\text{t})$**

# What is the goal of reinforcement learning?

maximize summation of rewards:  **$max\Sigma_{t=1}^T r(s_\text{t},a_\text{t})$**

But this is not a deterministic quantity!

What are the sources of variability?

1. the world is stochastic.
2. the car may not make the same decision every time.

**$$P(\tau) = P(s_\text{1}) \Pi_{t=1}^T \pi_\theta(a_\text{t}|s_\text{t})P(s_\text{t+1}|s_\text{t}, a_\text{t})$$**

The goal is to maximize expected sum of rewards under this equation of probabilities.

**$$max_\pi E_{\tau \sim P(\tau)}[r(s_\text{t},a_\text{t})] = max_\pi E_{\tau \sim P(\tau)}[r(s_\text{1},a_\text{1}) + r(s_\text{2},a_\text{2}) + \ldots + r(s_\text{T},a_\text{T})]$$**

# How do we evaluate how policy is good?

value function **$V^\pi(s)$** = future expected reward starting at **$s$** and following **$\pi$**

Q value function **$Q^\pi(s,a)$** = expected reward starting at **$s$** and taking action **$a$** following **$\pi$**


# Type of algorithms

1. **imitation learning**: mimic a policy that achieves high reward
2. **policy gradients**: directly differentiate the above objective
3. **Actor-critic**: estimate the values of the current policy and use it to make the policy better.
4. **Value-based**: estimate the values of the optional policy
5. **Model-based**: learn to model the dynamics and use it for planning or policy improvement

# Generative models for policies(approximating **$p(a|s)$**)

1. Mixture of Gaussians
