# Actor-Critic Methods

Some Useful Objects should be reminded:

1. **value function **$V^\pi(s)$**** - future expected rewards starting at **$s$** and following **$\pi$**
**$$V^\pi(s_t)=\sum_{t' = t}^T E_{s_{t'},a_{t'} \sim \pi_\theta} [r(s_{t'},a_{t'})|s_t]$$**
2. **Q function **$Q^\pi(s,a)$**** - expected rewards starting at **$s$** and taking **$a$** following **$\pi$**
**$$Q^\pi(s,a)=E_{s_{t'},r_{t'} \sim \pi_\theta} [r_{t'}|s_t,a_t]$$**
3. **advantage** - difference between **$Q$** and **$V$**: how much better it is to take action **$a$** than to follow policy **$\pi$** at state **$s$**
**$$A^\pi(s,a)=Q^\pi(s,a)-V^\pi(s)$$**

The Policy Gradient would be shown as followed with Q function:
$$\nabla_\theta J(\theta)=\frac{1}{N} \sum_{i=1}^N \sum_{t =1}^T \nabla_\theta \log \pi_\theta(a_{i, t}|s_{i, t})(Q(s_{i, t}, a_{i, t}) - b)$$
$$\nabla_\theta J(\theta)=\frac{1}{N} \sum_{i=1}^N \sum_{t =1}^T \nabla_\theta \log \pi_\theta(a_{i, t}|s_{i, t})(Q(s_{i, t}, a_{i, t}) - V^\pi(s_t))$$

$$\nabla_\theta J(\theta)=\frac{1}{N} \sum_{i=1}^N \sum_{t =1}^T \nabla_\theta \log \pi_\theta(a_{i, t}|s_{i, t})A^\pi(s_{i, t},a_{i, t})$$

$$\nabla_\theta J(\theta)=E_{s,a \sim \pi_\theta}[A^\pi(s_{i, t},a_{i, t})]$$

This is improving policy gradient.

This circle loop will be convoluted.
First, initialize the policy (randomly, with imitation learning, with heuristics)
Run policy to collect batch of data **$\to$** Fit model to estimate expected return **$\to$** Improve policy **$\to$** Repeat

$$A^\pi(s,a)=r(s_t, a_t)+V^\pi(s_{t+1})-V^\pi(s_t)$$

So, Base on the equations above, we just fit $V^\pi$

# How do we fit $V^\pi$?

## 1. Monte Carlo estimation

when only one original single sample estimate:
$$V^\pi(s_t) = \sum_{t'=t}^T r(s_{t'}, a_{t'})$$

when multiple-sample estimate(but can't reset the world):
$$V^\pi(s_t) = \frac{1}{N} \sum_{i=1}^N \sum_{t'=1}^T r(s_{t'},a_{t'})$$

if we trained the dataset, the dataset would be like below:
$$(s_{i, t}, \sum_{t'=t}^T r(s_{i, t'}, a_{i, t'}))$$
