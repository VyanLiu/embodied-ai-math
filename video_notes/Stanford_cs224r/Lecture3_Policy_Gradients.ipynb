# Policy Gradients

1. **Online RL Outline**: First Initialize the policy (randomly, with imitation learning, with heuristics) And then, run the loop, Run policy to collect batch of data **$to$** Improve policy using batch of data and the back to the start. Try this loop so hard to get better solutions.

This is the process that how to run the policy to get a rollout in the environment.
## Evaluate the RL objective

- **Objective Function**: The objective function in policy gradient methods is to maximize the expected return $$\theta^* = arg  max_\theta E_{\tau \sim p_\theta(\tau)}[\sum_{t=0}^{T-1} r(s_t, a_t)]$$, where $\tau$ is a trajectory, $\pi_\theta$ is the policy parameterized by $\theta$, and $r(s_t, a_t)$ is the reward at time step $t$.

$$J(\theta) = E_{\tau \sim p_\theta(\tau)}[\sum_{t=0}^{T-1} r(s_t, a_t)] = \frac{1}{N} \sum_{i=1}^{N} \sum_{t=0}^{T-1} r(s_t^i, a_t^i)$$

We could run the policy over and over time(N times), observe the resulting reward, and then compute the average over the observing reward. And that is an estimate of how good our policy is.

## How do we improve the policy after we've collected that data?  The definition of policy gradient

$$\nabla_\theta J(\theta) = E_{\tau \sim p_\theta(\tau)}[\sum_{t=0}^{T-1} \nabla_\theta \log P_\theta(\tau) r(s_t, a_t)]$$

**$$P_\theta(\tau) = P(s_\text{1}) \Pi_{t=1}^T \pi_\theta(a_\text{t}|s_\text{t})P(s_\text{t+1}|s_\text{t}, a_\text{t})$$**

**$$log P_\theta(\tau) = log P(s_\text{1}) + \sum_{t=1}^T log \pi_\theta(a_\text{t}|s_\text{t}) + log P(s_\text{t+1}|s_\text{t}, a_\text{t})$$**

$$\nabla_\theta log P_\theta(\tau) = \nabla_\theta log \pi_\theta(a_t|s_t)$$

$$\nabla_\theta J(\theta) = E_{\tau \sim p_\theta(\tau)}[\sum_{t=0}^{T-1} \nabla_\theta \log \pi_\theta(a_t|s_t) r(s_t, a_t)]$$

$$\nabla_\theta J(\theta) = \frac{1}{N} \sum_{i=1}^{N} [\sum_{t=0}^{T-1} \nabla_\theta \log \pi_\theta(a_t^i|s_t^i) r(s_t^i,a_t^i)]$$

$$\theta_{t+1} = \theta_t + \alpha \nabla_\theta J(\theta_t)$$

Full Algorithm:
1. sample **${\tau^{i}}$** from **$\pi_\theta(a_t|s_t)$**
2. **$\nabla_\theta J(\theta) = \sum_i(\sum_t \nabla_\theta \log \pi_\theta(a_t^i|s_t^i)) (\sum_t r(s_t^i,a_t^i))$**
3. **$\theta_{t+1} = \theta_t + \alpha \nabla_\theta J(\theta_t)$**

An interesting observation is that the policy gradient is the imitation gradient, but weighted by rewards.
Intuition:
    Increase the likelihood of actions that you took in high reward trajectory.
    Decrease the likelihood of actions that you took in low reward trajectory.

Reward is just weight, we can do anything as weights, such as cross attention, put image embeddings and text embeddings together and cross the trajectory as the weight or reward of the trajectory. And then softmax them and apply these weights to the actions itself.

## Improving the gradient (introducing baselines)

**$$\nabla_\theta J(\theta) = E_{\tau \sim p_\theta(\tau)}[\sum_{t=0}^{T-1} \nabla_\theta \log \pi_\theta(a_t|s_t) (r(s_t, a_t) - b(s_t))]$$**

**$$b(s_t) = E_{\tau \sim p_\theta(\tau)}[r(s_t, a_t)]$$**

If we subtract average rewards, we get negative gradients for below-average behavior.

Why can we add b to the gradient based on mathematical theorem?

**$$E[\nabla_\theta \log p_\theta(\tau)b] = \int p_\theta(\tau) \nabla_\theta \log p_\theta(\tau)bd\tau = \int \nabla_\theta p_\theta(\tau) bd\tau = b \nabla_\theta \int p_\theta(\tau) d\tau = b \nabla_\theta 1 = 0$$**
